```python
import pandas as pd
import re
from langdetect import detect
from deep_translator import GoogleTranslator
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('punkt')
nltk.download('wordnet')

translator = GoogleTranslator(source='auto', target='en')
lemmatizer = WordNetLemmatizer()

# Example: load one JSON (adjust path after download)
example_file = "../data/raw/miller_speeches/some_speech.json"  # change to real file
# df = pd.read_json(example_file, lines=True)   # later loop over many

# For testing – pretend we have text
sample_text = "This is a test! http://example.com @user #tag RT"

def clean(text):
    text = re.sub(r'http\S+|@\w+|RT|#\S+', '', str(text))
    text = re.sub(r'\s+', ' ', text).strip()
    return text

print("Cleaned:", clean(sample_text))

# Translation example
def translate_if_needed(text):
    if len(text) < 30: return text
    try:
        if detect(text) != 'en':
            return translator.translate(text[:4500])  # API limit
        return text
    except:
        return text

print("Translated test:", translate_if_needed("Esto es una prueba en español."))

# Lemmatize
def lemmatize(text):
    tokens = nltk.word_tokenize(text.lower())
    return " ".join(lemmatizer.lemmatize(t) for t in tokens if t.isalpha())

print("Lemmatized:", lemmatize("running tests better"))
```
